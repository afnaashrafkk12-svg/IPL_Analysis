```python
import sqlite3
import pandas as pd 
con = sqlite3.connect('../data/raw/ipl.db')
```

```python
def query(sql):
   return pd.read_sql(sql,con)
```

```python
def run_sql_file(path):
    with open(path,'r')as f:
        script = f.read()
    con.executescript(script)
    con.commit()
    print('ran',path)
```

```python
query('SELECT COUNT(*)FROM matches;')
```

```python
query('''SELECT COUNT (*)AS looks_empty
FROM matches
WHERE player_of_match = '';''')
```

```python
query('''SELECT CoUnT(*)AS found_by_is_null
FROM matches
WHERE player_of_match IS NULL;''')
```

```python
query('''
SELECT length (bowler_type)AS len,COUNT(*) AS rows
fROM deliveries
WHERE bowler_type IS NULL OR TRIM (bowler_type)=''
GROUP BY len;''')
```

```python
query('''
SELECT TRIM( '  hello  ' ) AS trimmed,
    LENGTH( '  hello  ') AS before,
    LENGTH(TRIM(' hello ')) AS after;''')
```

```python
query('''
SELECT NULLIF('a', 'a') AS same,
       NULLIF('a', 'b') AS different,
       NULLIF('', '') AS empty;''')
```
