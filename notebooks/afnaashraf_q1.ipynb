QUESTION-2

This query shows all the different seasons from the matches table.

It does not show the same season more than once

```python
q('''
SELECT DISTINCT season 
FROM matches;
''')
```

```python
q('''
SELECT match_id,city
FROM matches
WHERE city='Mumbai';
''')
```

```python
q('''SELECT COUNT(*) AS total_matches
FROM matches;
''')
```
